```python
# Array of Edges

n = 10

A = [[0,1], [1,2], [1,4], [2,3], [2,5], [3,7], [4,8], [5,6], [7,10], [8,9]]

print(A)
```

```python
# Adjency Matrix

M = []

for i in range(n+1):
    M.append([0]*(n+1))
    
for u,v in A:
    M[u][v] = 1

M
```

```python
# Adjency List

from collections import defaultdict

D = defaultdict(list)

for u,v in A:
    D[u].append(v)
    
D
```

```python
# BFS

source = 0

from collections import deque

seen = set()
q=deque()
seen.add(source)
q.append(source)

while q:
    node = q.popleft()
    print(node)
    for nei in D[node]:
        if nei not in seen:
            seen.add(nei)
            q.append(nei)
            
```

```python
# DFS Recursive

source = 0

def recursive_dfs(node):
    print(node)
    for nei in D[node]:
        if nei not in seen:
            seen.add(nei)
            recursive_dfs(nei)

seen = set()
seen.add(source)
recursive_dfs(source)
```
