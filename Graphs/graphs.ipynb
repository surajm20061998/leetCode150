```python
# Union Find

# 11. How to recognize Union Find problems

# Think about Union Find whenever the problem involves:

# * Groups of connected elements
# * Merging two groups
# * Checking whether two elements belong to the same group
# * Detecting whether an undirected edge creates a cycle
# * Maintaining connectivity as edges are added

numberOfNodes = 11

edges = [
    [0, 1],
    [1, 2],
    [2, 3],
    [3, 7],
    [4, 8],
    [5, 6],
    [7, 10],
    [8, 9]
]

print(edges)

class Solution:
    def countComponents(self, n, edges):

        parent = list(range(n))
        size = [1] * n

        def find(x):
            while x != parent[x]:
                parent[x] = parent[parent[x]]
                x = parent[x]
            return x

        def union(a, b):
            rootA = find(a)
            rootB = find(b)

            if rootA == rootB:
                return False

            if size[rootA] < size[rootB]:
                rootA, rootB = rootB, rootA

            parent[rootB] = rootA
            size[rootA] += size[rootB]

            return True

        components = n

        for u, v in edges:
            if union(u, v):
                components -= 1

        return components
sol = Solution()
print(sol.countComponents(numberOfNodes,edges))



```
